```python
import pandas as pd

df_metricas = pd.read_csv("data.csv", delimiter= ";", encoding="utf-8")
display(df_metricas)
```

```python
display(df_metricas.info())
#pegar dados (ver informações dos mesmos)

print(list(df_metricas.columns))
```

```python
print(df_metricas["cpu_frequency"])

df_metricas_analisadas = df_metricas

df_metricas_analisadas["timestamp"] = pd.to_datetime(df_metricas_analisadas["timestamp"], format="%Y-%m-%d %H:%M:%S.%f")
```

```python
df_metricas_analisadas["timestamp"].info()
#agora ta datetime

df_metricas_analisadas["user"] = df_metricas_analisadas["user"].str.upper()
df_metricas_analisadas["user"]

df_metricas_analisadas = df_metricas_analisadas.drop_duplicates(subset=["timestamp"], keep="first")

df_metricas_analisadas["timestamp"] = df_metricas_analisadas["timestamp"].dt.floor('s')

df_metricas_analisadas = df_metricas_analisadas.fillna(0)

display(df_metricas_analisadas)
```

```python
display(df_metricas_analisadas)
```

```python
print(df_metricas_analisadas["cpu_percent"].mean())
display(df_metricas_analisadas["cpu_percent"])
print(df_metricas_analisadas["cpu_percent"].max())

cpu_alta = df_metricas_analisadas[df_metricas_analisadas["ram_percent"] > 95.0]
display(cpu_alta)

display(cpu_alta["timestamp"].dt.day_name())
#as duas no sabado
```

```python
#media por dia da semana
media_cpu_por_dia = df_metricas_analisadas.groupby(df_metricas_analisadas["timestamp"].dt.day_name())["cpu_percent"].mean()

display(media_cpu_por_dia)

dados_do_sabado = df_metricas_analisadas[df_metricas_analisadas["timestamp"].dt.weekday == 5]

print("media sabado")
print(dados_do_sabado["cpu_percent"].mean())
```

```python

df_semtexto = df_metricas_analisadas.drop(columns=["user"])
ultima = df_semtexto["timestamp"].max()
uma_hora =  df_semtexto[df_semtexto["timestamp"] >= (ultima - pd.Timedelta(hours=1))]
um_dia =  df_semtexto[df_semtexto["timestamp"] >= (ultima - pd.Timedelta(hours=24))]

display(df_semtexto)

df_media_ultima_hora = uma_hora.groupby(uma_hora["timestamp"].dt.floor('h'), "user").mean()
df_media_ultimo_dia = um_dia.groupby(um_dia["timestamp"].dt.floor('h'), "user").mean()

display(df_media_ultima_hora)
display(df_media_ultimo_dia)
```
